```python
from statsbombpy import sb
import pandas as pd

# Bring in Mens World Cup 2022
matches = sb.matches(competition_id = 43, season_id = 106) 
shots = []
for mid in matches.match_id:
    ev = sb.events(match_id = mid)
    shots.append(ev[ev.type == "Shot"])

shots = pd.concat(shots)
shots.to_parquet("../data/shots_wc2022.parquet")
```

```python
# Gives overall dimensions of shots
shots.shape
```

```python
# Gives outcome of each shot
shots.shot_outcome.value_counts()
```

```python
# Show foot vs head shots
display(shots.shot_body_part.value_counts())

# Shows if shot was in open play, free kick, or penalty
display(shots.shot_type.value_counts())
```
